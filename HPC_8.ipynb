```python
#assignment-1
import time
import multiprocessing

def compute(n):
    s = 0
    for i in range(n):
        s += i
    return s

def run_test(processes):
    n = 10000000
    start = time.time()

    with multiprocessing.Pool(processes) as pool:
        pool.map(compute, [n//processes]*processes)

    end = time.time()
    return end - start

if __name__ == "__main__":
    serial_time = run_test(1)

    for p in [1,2,4,8]:
        t = run_test(p)
        speedup = serial_time / t
        efficiency = speedup / p

        print(f"Processes: {p}")
        print(f"Execution Time: {t:.4f}")
        print(f"Speedup: {speedup:.4f}")
        print(f"Efficiency: {efficiency:.4f}\n")
```

```python
#asignment-3
import multiprocessing
import time
import os

def task(n):
    start = time.time()

    total = 0
    for i in range(n):
        total += i*i

    end = time.time()
    print(f"Process {os.getpid()} time: {end-start:.4f}")

if __name__ == "__main__":
    work = [10000000, 20000000, 5000000, 30000000]

    processes = []
    for w in work:
        p = multiprocessing.Process(target=task, args=(w,))
        p.start()
        processes.append(p)

    for p in processes:
        p.join()
```

```python
#assignment-4
import multiprocessing

def update_array(arr, start, end):
    for i in range(start, end):
        arr[i] += 1

if __name__ == "__main__":
    size = 1000000
    arr = multiprocessing.Array('i', size)

    p1 = multiprocessing.Process(target=update_array, args=(arr,0,size//2))
    p2 = multiprocessing.Process(target=update_array, args=(arr,size//2,size))

    p1.start()
    p2.start()

    p1.join()
    p2.join()

    print("Array updated successfully")
```

```python
#assignment-5
import multiprocessing

def worker(counter, lock):
    for _ in range(100000):
        lock.acquire()
        counter.value += 1
        lock.release()

if __name__ == "__main__":
    counter = multiprocessing.Value('i',0)
    lock = multiprocessing.Lock()

    processes = []

    for _ in range(4):
        p = multiprocessing.Process(target=worker, args=(counter,lock))
        p.start()
        processes.append(p)

    for p in processes:
        p.join()

    print("Final Counter:", counter.value)
```
